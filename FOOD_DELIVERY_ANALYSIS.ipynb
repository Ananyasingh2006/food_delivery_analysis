```python
print("========== FOOD DELIVERY ANALYSIS ==========")

# Total orders
print("Total Orders:", len(df))

# Average order cost
print("Average Order Cost:", round(df['cost_of_the_order'].mean(), 2))

# Average delivery time
print("Average Delivery Time:",
      round(df['delivery_time'].mean(), 2), "minutes")

# Average preparation time
print("Average Preparation Time:",
      round(df['food_preparation_time'].mean(), 2), "minutes")

# Most popular cuisine
print("Most Popular Cuisine:",
      df['cuisine_type'].value_counts().idxmax())

# Most ordered restaurant
print("Most Ordered Restaurant:",
      df['restaurant_name'].value_counts().idxmax())

print("============================================")
```
